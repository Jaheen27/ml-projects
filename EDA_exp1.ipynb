```python
# creating new column movie age
from datetime import datetime

mv['Movie_Age']=datetime.now().year-mv['Year']
mv.head()
```

```python
mv.drop('overview',axis=1,inplace=True)
```

```python
# Distribution Plot of Movie Ratings

plt.figure(figsize=(10, 6))

sns.histplot(mv['rating'].dropna(),
             kde=True,
             color='purple')

plt.title("Distribution of Movie Ratings")
plt.xlabel("Rating")
plt.ylabel("Frequency")

plt.show()
```

```python
# Scatter Plot: Vote Count vs Rating

plt.figure(figsize=(8,5))

plt.scatter(mv['vote_count'],
            mv['rating'],
            alpha=0.5)

plt.xlabel("Vote Count")
plt.ylabel("Movie Rating")
plt.title("Vote Count vs Movie Rating")

plt.show()
```
